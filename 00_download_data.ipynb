```python
import requests, os

url = "https://nimbus.niu.edu/DATA/SDHI/SDHI_Precip_Supporting_data.tar.gz"
os.makedirs("../data", exist_ok=True)
fname = "../test_data/SDHI_Precip_Supporting_data.tar.gz"

if not os.path.exists(fname):
    print("Downloading...")
    r = requests.get(url, stream=True)    # we stream since the file is large
    r.raise_for_status()
    with open(fname, 'wb') as f:
        for chunk in r.iter_content(chunk_size=8192): f.write(chunk)
    print("Done.")
else:
    print("Already downloaded.")
```

```python
import tarfile

print("Extracting...")
with tarfile.open(fname, "r:gz") as tar:
    tar.extractall("../test_data")
print("Done.")
```
